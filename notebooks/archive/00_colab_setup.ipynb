```python
from pathlib import Path

DRIVE_ROOT = Path('/content/drive/MyDrive/Vietnamese_VQA_NLE')
ARCHIVE = DRIVE_ROOT / 'datasets/images/coco_smoke_500.zip'
ANNOTATIONS = DRIVE_ROOT / 'datasets/processed/smoke_500.jsonl'
WORK_ROOT = Path('/content/vqa_work/coco')

assert ARCHIVE.is_file(), ARCHIVE
assert ANNOTATIONS.is_file(), ANNOTATIONS
WORK_ROOT.mkdir(parents=True, exist_ok=True)
```

```python
import hashlib
import zipfile

EXPECTED_SHA256 = '4e0fce747dee9a1beaf36e2617deae6582115020d7c219f315525e329a4024f9'
digest = hashlib.sha256(ARCHIVE.read_bytes()).hexdigest()
assert digest == EXPECTED_SHA256, (digest, EXPECTED_SHA256)
with zipfile.ZipFile(ARCHIVE) as archive:
    archive.extractall(WORK_ROOT)
```

```python
import json

from PIL import Image

records = [json.loads(line) for line in ANNOTATIONS.read_text(encoding='utf-8').splitlines()]
image_paths = sorted((WORK_ROOT / 'images').glob('*.jpg'))
assert len(image_paths) == 500, len(image_paths)
assert len(records) == 588, len(records)
for path in image_paths:
    with Image.open(path) as image:
        image.verify()
print(f'Ready: {len(image_paths)} images, {len(records)} QA pairs')
```
